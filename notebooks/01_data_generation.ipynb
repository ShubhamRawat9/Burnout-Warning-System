```python
import pandas as pd
import numpy as np

# Set random seed — ensures same results every time you run
np.random.seed(42)

print("Libraries imported")
print(f"Pandas: {pd.__version__}")
print(f"NumPy:  {np.__version__}")
```

```python
# parameters
N_EMPLOYEES = 100   # number of simulated employees
N_WEEKS     = 16    # weeks of tracking per employee

print(f"Parameters set")
print(f"Employees : {N_EMPLOYEES}")
print(f"Weeks     : {N_WEEKS}")
print(f"Total rows: {N_EMPLOYEES * N_WEEKS}")
```

```python
records = []

for emp_id in range(1, N_EMPLOYEES + 1):
    for week in range(1, N_WEEKS + 1):
        if week <= 6:
            phase            = "Low"
            completion_rate  = np.random.uniform(0.85, 1.0)
            focus_sessions   = np.random.randint(4, 7)
            focus_duration   = np.random.uniform(60, 120)
            context_switches = np.random.randint(2, 6)
            idle_time        = np.random.uniform(0, 40)
            work_hours       = np.random.uniform(7, 9)
            after_hours      = np.random.randint(0, 2)
            breaks_taken     = np.random.randint(3, 6)
            weekend_work     = np.random.randint(0, 1)
            error_rate       = np.random.uniform(0.02, 0.08)
            delays           = np.random.randint(0, 2)
            revisions        = np.random.randint(0, 2)
            fatigue_score    = np.random.uniform(1, 3)

        elif week <= 11:
            phase            = "Medium"
            completion_rate  = np.random.uniform(0.65, 0.84)
            focus_sessions   = np.random.randint(2, 5)
            focus_duration   = np.random.uniform(30, 60)
            context_switches = np.random.randint(6, 14)
            idle_time        = np.random.uniform(40, 100)
            work_hours       = np.random.uniform(9, 12)
            after_hours      = np.random.randint(2, 5)
            breaks_taken     = np.random.randint(1, 3)
            weekend_work     = np.random.randint(0, 2)
            error_rate       = np.random.uniform(0.10, 0.25)
            delays           = np.random.randint(2, 4)
            revisions        = np.random.randint(2, 5)
            fatigue_score    = np.random.uniform(4, 6)

        else:
            phase            = "High"
            completion_rate  = np.random.uniform(0.30, 0.64)
            focus_sessions   = np.random.randint(1, 3)
            focus_duration   = np.random.uniform(10, 30)
            context_switches = np.random.randint(14, 22)
            idle_time        = np.random.uniform(100, 180)
            work_hours       = np.random.uniform(10, 14)
            after_hours      = np.random.randint(5, 8)
            breaks_taken     = np.random.randint(0, 2)
            weekend_work     = np.random.randint(1, 3)
            error_rate       = np.random.uniform(0.25, 0.50)
            delays           = np.random.randint(4, 7)
            revisions        = np.random.randint(5, 9)
            fatigue_score    = np.random.uniform(7, 10)

        tasks_planned   = np.random.randint(8, 18)
        tasks_completed = int(tasks_planned * completion_rate)
        rework_count    = int(tasks_completed * error_rate)

        records.append({
            "employee_id"         : f"E{emp_id:03d}",
            "week_number"         : week,
            "burnout_label"       : phase,
            "tasks_planned"       : tasks_planned,
            "tasks_completed"     : tasks_completed,
            "completion_rate"     : round(completion_rate, 3),
            "rework_count"        : rework_count,
            "focus_sessions"      : focus_sessions,
            "avg_focus_duration"  : round(focus_duration, 1),
            "context_switches"    : context_switches,
            "idle_time_mins"      : round(idle_time, 1),
            "work_hours_per_day"  : round(work_hours, 1),
            "after_hours_logins"  : after_hours,
            "breaks_taken"        : breaks_taken,
            "weekend_work_days"   : weekend_work,
            "error_rate"          : round(error_rate, 3),
            "submission_delays"   : delays,
            "revision_requests"   : revisions,
            "mental_fatigue_score": round(fatigue_score, 1),
        })

df = pd.DataFrame(records)
print(f" Dataset generated")
print(f"Shape : {df.shape}")
print(f"\nLabel distribution:")
print(df["burnout_label"].value_counts())
```
